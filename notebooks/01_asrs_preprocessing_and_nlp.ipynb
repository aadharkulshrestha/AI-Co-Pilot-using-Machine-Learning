# NASA ASRS Incident Report Preprocessing & NLP Feature Extraction
### AI Co-Pilot: Predictive Pilot Decision-Making & Flight Risk Assessment

This notebook demonstrates the extraction, tokenization, and semantic vectorization of NASA Aviation Safety Reporting System (ASRS) pilot incident narratives.

```python
import re
import json
import numpy as np
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import TruncatedSVD
import matplotlib.pyplot as plt
```

## 1. Load NASA ASRS Incident Dataset

```python
# Load sample ASRS records generated by AI Co-Pilot
from backend.app.ml.dataset_generator import generate_full_synthetic_dataset
_, _, _, asrs_records = generate_full_synthetic_dataset(num_samples=200)
df_asrs = pd.DataFrame(asrs_records)
print(f"Total ASRS records loaded: {len(df_asrs)}")
df_asrs.head(3)
```

## 2. Text Cleaning & Tokenization Pipeline

```python
def clean_aviation_narrative(text):
    text = text.lower()
    text = re.sub(r'[^a-zA-Z0-9\s]', ' ', text)
    text = re.sub(r'\s+', ' ', text).strip()
    return text

df_asrs['cleaned_narrative'] = df_asrs['narrative'].apply(clean_aviation_narrative)
print("Sample Cleaned Narrative:")
print(df_asrs['cleaned_narrative'].iloc[0][:180] + '...')
```

## 3. TF-IDF & Semantic Feature Extraction

```python
tfidf = TfidfVectorizer(max_features=100, stop_words='english')
X_tfidf = tfidf.fit_transform(df_asrs['cleaned_narrative']).toarray()
print(f"TF-IDF Feature Matrix Shape: {X_tfidf.shape}")
print(f"Top vocabulary tokens: {list(tfidf.vocabulary_.keys())[:10]}")
```
