# Explainable AI: SHAP Waterfall & Attention Map Visualizations
### AI Co-Pilot: Model Interpretability for Cockpit Decision Support

This notebook demonstrates how SHAP values and LSTM temporal attention maps explain why specific pilot actions and risk scores were generated.

```python
from backend.app.ml.explainer import AviationExplainer

explainer = AviationExplainer()
sample_telemetry = {
    'altitude': 1100.0,
    'airspeed': 195.0,
    'vertical_rate': -3950.0,
    'pitch': -7.0,
    'roll': 3.0,
    'wind_speed': 20.0,
    'throttle': 45.0
}

res = explainer.explain_telemetry_state(
    raw_telemetry=sample_telemetry,
    predicted_action='Increase Altitude & Thrust',
    detected_event='Excessive Descent Rate',
    risk_score=94.0
)

print("Top Contributing Factors:", res['top_factors'])
print("\nNatural Language Explanation:")
print(res['natural_language_explanation'])
print("\nSHAP Waterfall Factors:")
for f in res['shap_waterfall']:
    print(f" - {f['feature']} ({f['val_str']}): impact +{f['impact']:.1f}%")
```
